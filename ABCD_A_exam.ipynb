```python
import os
import glob
from pathlib import Path

def find_data_yaml():
    """data.yaml 파일을 찾는 함수"""

    # 방법 1: 현재 작업 디렉토리에서 찾기
    current_dir = os.getcwd()
    yaml_path = os.path.join(current_dir, 'data.yaml')

    if os.path.exists(yaml_path):
        print(f"✓ 현재 디렉토리에서 발견: {yaml_path}")
        return yaml_path

    # 방법 2: 재귀적으로 모든 하위 디렉토리 검색
    print("현재 디렉토리에서 data.yaml을 찾는 중...")
    for root, dirs, files in os.walk(current_dir):
        if 'data.yaml' in files:
            found_path = os.path.join(root, 'data.yaml')
            print(f"✓ 발견: {found_path}")
            return found_path

    # 방법 3: glob을 사용한 패턴 매칭
    print("glob 패턴으로 검색 중...")
    yaml_files = glob.glob('**/data.yaml', recursive=True)
    if yaml_files:
        for file in yaml_files:
            print(f"✓ 발견: {os.path.abspath(file)}")
        return os.path.abspath(yaml_files[0])

    # 방법 4: 일반적인 위치들 확인
    common_paths = [
        '/content/data.yaml',
        '/content/drive/MyDrive/data.yaml',
        '/content/datasets/data.yaml',
        './datasets/data.yaml',
        './data/data.yaml'
    ]

    print("일반적인 위치에서 검색 중...")
    for path in common_paths:
        if os.path.exists(path):
            print(f"✓ 발견: {path}")
            return path

    print("❌ data.yaml 파일을 찾을 수 없습니다.")
    return None

# 실행
yaml_location = find_data_yaml()

# 추가: 모든 .yaml 파일 찾기
print("\n=== 모든 .yaml 파일 검색 ===")
all_yaml_files = glob.glob('**/*.yaml', recursive=True)
all_yml_files = glob.glob('**/*.yml', recursive=True)

yaml_files = all_yaml_files + all_yml_files
if yaml_files:
    print("발견된 YAML 파일들:")
    for file in yaml_files:
        print(f"  - {os.path.abspath(file)}")
else:
    print("YAML 파일이 없습니다.")

# 현재 디렉토리 구조 확인
print(f"\n=== 현재 작업 디렉토리 ===")
print(f"위치: {os.getcwd()}")
print("\n디렉토리 내용:")
```

```python
import cv2

def draw_enhanced_results(image_path, results):
    """세그멘테이션 마스크를 이미지 위에 시각화"""
    # 이미지 로드
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 클래스 색상 정의
    class_names = ['face', 'open_eyes', 'close_eyes']
    colors = [(255, 0, 0), (0, 255, 0), (0, 0, 255)]

    for result in results:
        masks = result.masks  # 마스크 객체

        if masks is not None and masks.data is not None:
            mask_data = masks.data.cpu().numpy()  # (N, H, W)
            cls_list = result.boxes.cls.cpu().numpy().astype(int)  # 클래스 인덱스 리스트
            conf_list = result.boxes.conf.cpu().numpy()  # 신뢰도

            for i in range(len(mask_data)):
                mask = mask_data[i]
                cls_id = cls_list[i]
                conf = conf_list[i]

                # 마스크 색상 및 클래스 이름
                color = colors[cls_id % len(colors)]
                class_name = class_names[cls_id] if cls_id < len(class_names) else f'class_{cls_id}'
                label = f'{class_name}: {conf:.2f}'

                # 마스크를 컬러로 적용
                colored_mask = np.zeros_like(img_rgb, dtype=np.uint8)
                for c in range(3):
                    colored_mask[:, :, c] = mask * color[c]

                # 반투명하게 overlay
                img_rgb = cv2.addWeighted(img_rgb, 1.0, colored_mask, 0.4, 0)

                # 마스크 경계 추출하여 텍스트 위치 추정
                contours, _ = cv2.findContours((mask * 255).astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                if contours:
                    x, y, w, h = cv2.boundingRect(contours[0])
                    # 라벨 텍스트
                    (text_width, text_height), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.7, 2)
                    cv2.rectangle(img_rgb, (x, y - text_height - 10), (x + text_width, y), color, -1)
                    cv2.putText(img_rgb, label, (x, y - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)

    return img_rgb
```

```python
import cv2
import numpy as np

def draw_enhanced_results(image_path, results):
    img = cv2.imread(image_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w, _ = img_rgb.shape

    class_names = ['face', 'open_eyes', 'close_eyes']

    # 🟩 클래스별 고정 색상 (BGR 형식)
    class_colors = {
        0: (255, 0, 0),     # face → 빨강
        1: (0, 255, 0),     # open_eyes → 초록
        2: (0, 0, 255),     # close_eyes → 파랑
    }

    for r in results:
        boxes = r.boxes
        masks = r.masks

        if masks is not None and boxes is not None:
            mask_data = masks.data.cpu().numpy()  # (N, H, W)
            cls_list = boxes.cls.cpu().numpy().astype(int)  # class index
            conf_list = boxes.conf.cpu().numpy()  # confidence

            for i, mask in enumerate(mask_data):
                # 마스크 리사이즈
                mask_resized = cv2.resize(mask, (w, h), interpolation=cv2.INTER_NEAREST)

                # 클래스 색상 지정 (기본값은 회색)
                cls_id = cls_list[i]
                color = class_colors.get(cls_id, (128, 128, 128))  # BGR
                conf = conf_list[i]
                label = f"{class_names[cls_id] if cls_id < len(class_names) else f'class_{cls_id}'}: {conf:.2f}"

                # 컬러 마스크 생성
                colored_mask = np.zeros_like(img_rgb, dtype=np.uint8)
                for c in range(3):
                    colored_mask[:, :, c] = mask_resized * color[2 - c]  # RGB로 맞추기

                # 마스크 오버레이
                img_rgb = cv2.addWeighted(img_rgb, 1.0, colored_mask, 0.5, 0)

                # 라벨 위치 (마스크 윤곽 기반)
                contours, _ = cv2.findContours((mask_resized * 255).astype(np.uint8), cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                if contours:
                    x, y, w_box, h_box = cv2.boundingRect(contours[0])
                    (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 1)
                    cv2.rectangle(img_rgb, (x, y - th - 5), (x + tw, y), color, -1)
                    cv2.putText(img_rgb, label, (x, y - 2), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,255), 1)

    return cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)
```

```python
#동영상으로 만들기

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

# YOLO 예측 + 시각화 함수 (콜라보용)
image_dir = '/content/dataset/test/images/'
test_images = sorted([img for img in os.listdir(image_dir) if img.lower().endswith(('.png', '.jpg', '.jpeg'))])

frames = []

for i, img_name in enumerate(test_images[:427]):  # 처음 5개만
    img_path = os.path.join(image_dir, img_name)

    # 예측 실행
    result = model.predict(img_path, conf=0.5, verbose=False)

    # 향상된 시각화
    enhanced_img = draw_enhanced_results(img_path, result)

    # RGB -> BGR (OpenCV용)
    bgr_img = cv2.cvtColor(np.array(enhanced_img), cv2.COLOR_RGB2BGR)

    # 모든 프레임 사이즈를 동일하게 유지
    bgr_img = cv2.resize(bgr_img, (1280, 720))

    frames.append(bgr_img)

# 🔄 동영상 생성
video_path = "/content/detection_results.mp4"
fps = 5  # 0.5초 간격 = 2프레임/초
height, width, _ = frames[0].shape
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 코덱 설정

out = cv2.VideoWriter(video_path, fourcc, fps, (width, height))

for frame in frames:
    out.write(frame)

out.release()
print(f"✅ 영상 저장 완료: {video_path}")

# Colab에서 다운로드 링크 제공
from google.colab import files
files.download(video_path)
```
