```python
import math

theta = math.radians(37)

v0 = 100
y0 = 140
v0x = v0 * math.cos(theta)
v0y = v0 * math.sin(theta)
g = 9.8
```

```python
# (a)
# 0 = 140 + v0y*t - 1/2*9.8*t^2 이므로
t = (-v0y + math.sqrt(v0y**2 + 2 * g * y0)) / g

print(f"지면에 도달하는 시간: {t:.2f} 초")
```

```python
# (b)
x = v0x * t
print(f"수평 거리: {x:.2f} m")
```

```python
# (c)
vx = v0x
vy = v0y - g * t

print(f"수평 속도 vx = {vx:.2f} m/s")
print(f"수직 속도 vy = {vy:.2f} m/s")
```

```python
# (d)
v = math.sqrt(vx**2 + vy**2)

print(f"지면 충돌 직전의 속도 크기 = {v:.2f} m/s")
```

```python
# (e)
theta_prime = math.degrees(math.atan(vy / vx))

print(f"속도 벡터가 수평면과 이루는 각도 = {theta_prime:.2f}°")
```
